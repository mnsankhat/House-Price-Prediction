```python
import numpy as np

def convert_amount(amount):
    if isinstance(amount, str):
        amount = amount.strip()
        if 'Cr' in amount:
            return float(amount.replace('Cr', '').strip()) * 1e7
        elif 'Lac' in amount:
            return float(amount.replace('Lac', '').strip()) * 1e5
    return np.nan
```

```python
df['Amount_numeric']= df['Amount(in rupees)'].apply(convert_amount)
```

```python
#  now work on title columns

import re

def  extract_bhk(title):
  if isinstance(title,str):
    pattern = r'(\d+)\s*BHK'
    match = re.search(pattern,title)
    if match:
      return int(match.group(1))
  return None
```

```python
df['BHK']= df['Title'].apply(extract_bhk)
```

```python
def extract_area(area_str):
    if isinstance(area_str, str):
        area_str = area_str.split()[0]
        return float(area_str)
    return np.nan
```

```python
df["Carpet_Area_sqft"] = df["Carpet Area"].apply(extract_area)

```

```python
df.drop(columns="Carpet Area", inplace=True)
```

```python
def extract_floor(value):
    if not isinstance(value, str):
        return None, None

    value = value.lower()

    # total floors
    if "out of" in value:
        total = value.split("out of")[1].strip()
        total = float(total)
    else:
        total = None

    # current floor
    if "lower basement" in value:
        current = -1
    elif "upper basement" in value:
        current = 0
    elif "ground" in value:
        current = 0
    else:
        try:
            current = float(value.split("out of")[0].strip())
        except:
            current = None

    return current, total
```

```python
df[["Current_Floor", "Total_Floors"]] = df["Floor"].apply(
    lambda x: pd.Series(extract_floor(x))
)
```

```python
df["Current_Floor"].fillna(df["Current_Floor"].median(), inplace=True)
df["Total_Floors"].fillna(df["Total_Floors"].median(), inplace=True)
```

```python
df.drop(columns=["Floor"], inplace=True)
```
